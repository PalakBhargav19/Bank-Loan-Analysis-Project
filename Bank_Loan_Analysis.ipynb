## Exploratory Data Analysis (EDA)
In this section, we will analyze different features of the loan dataset using visualizations to identify patterns and trends.

```python
loan_status = df["loan_status"].value_counts()

plt.figure(figsize=(6,4))
plt.bar(loan_status.index, loan_status.values)

plt.title("Loan Status Distribution")
plt.xlabel("Loan Status")
plt.ylabel("Number of Applications")

plt.show()
```

### Business Insight

- The majority of loan applications are approved.
- Rejected applications are lower than approved applications.
- The bank has a healthy loan approval rate.

## Education vs Loan Status

This visualization shows how loan approval varies based on the applicant's education level.


```python
education_status = pd.crosstab(df["education"], df["loan_status"])

education_status.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Education vs Loan Status")
plt.xlabel("Education")
plt.ylabel("Number of Applicants")
plt.xticks(rotation=0)
plt.show()
```

### Observation

- Graduate applicants have slightly more loan approvals than Non-Graduate applicants.
- In both education categories, approved loans are higher than rejected loans.
- Education may influence loan approval, but it is not the only deciding factor.

## Income vs Loan Amount

This scatter plot helps analyze the relationship between applicant income and the requested loan amount.

```python
plt.figure(figsize=(8,5))

plt.scatter(
    df["income_annum"],
    df["loan_amount"],
    alpha=0.6
)

plt.title("Income vs Loan Amount")
plt.xlabel("Annual Income")
plt.ylabel("Loan Amount")

plt.show()
```

### Observation

- Applicants with higher annual income generally apply for higher loan amounts.
- There is a positive relationship between income and loan amount.
- Some exceptions exist where applicants with similar incomes requested different loan amounts.

## CIBIL Score vs Loan Status

This chart compares the average CIBIL score for approved and rejected loan applications.

```python
cibil_status = df.groupby("loan_status")["cibil_score"].mean()

plt.figure(figsize=(6,4))
plt.bar(cibil_status.index, cibil_status.values)

plt.title("Average CIBIL Score by Loan Status")
plt.xlabel("Loan Status")
plt.ylabel("Average CIBIL Score")

plt.show()
```

### Observation

- Approved applicants have a higher average CIBIL score than rejected applicants.
- A good CIBIL score increases the chances of loan approval.
- Credit score is an important factor in the loan approval process.

## Correlation Heatmap

This heatmap shows the correlation between numerical features in the dataset. It helps identify which variables are positively or negatively related.

```python
numeric_df = df.select_dtypes(include=["int64", "float64"])

correlation = numeric_df.corr()

plt.figure(figsize=(10,6))

plt.imshow(correlation, cmap="coolwarm", aspect="auto")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns, rotation=90)
plt.yticks(range(len(correlation.columns)), correlation.columns)

plt.title("Correlation Heatmap")

plt.tight_layout()
plt.show()
```

### Observation

- The heatmap shows the correlation between numerical variables.
- Dark red indicates a strong positive correlation, while dark blue indicates a strong negative correlation.
- Features with higher positive correlation tend to increase together.
- This analysis helps identify important factors that influence loan approval.

```python
plt.figure(figsize=(8,5))

plt.boxplot(df["loan_amount"])

plt.title("Loan Amount Distribution")
plt.ylabel("Loan Amount")

plt.show()
```

## Observation

- The box plot helps identify outliers in loan amount.
- Most applicants requested loan amounts within a common range.
- A few applicants requested significantly higher loan amounts.

```python
loan_status = df["loan_status"].value_counts()

plt.figure(figsize=(6,6))

plt.pie(
    loan_status,
    labels=loan_status.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Loan Status Distribution")
plt.show()
```
